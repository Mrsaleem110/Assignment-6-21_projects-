2. Using cls
Assignment:
Create a class Counter that keeps track of how many objects have been created. Use a class variable and a class method with cls to manage and display the **count**

```python
class Counter:
    # Class variable to track number of objects created
    count = 0

    def __init__(self):
        # Increment the count when a new object is created
        Counter.count += 1

    @classmethod
    def get_count(cls):
        # Return the current count of instances
        return cls.count

# Example usage:
a = Counter()
b = Counter()
c = Counter()
e = Counter()
f = Counter()


print("Number of Counter instances created:", Counter.get_count())
```

8. The super() Function
Assignment:
Create a class Person with a constructor that sets the name. Inherit a class Teacher from it, add a subject field, and use super() to call the base class constructor.**bold text**

```python
# Base class
class Person:
    def __init__(self, name):
        self.name = name  # Set the name

    def display(self):
        print(f"Name: {self.name}")

# Derived class
class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)  # Calling the constructor of the base class Person
        self.subject = subject  # Adding subject field for the Teacher class

    def display(self):
        super().display()  # Calling the display method of the base class
        print(f"Subject: {self.subject}")

# Create an object of Teacher
teacher1 = Teacher("Muhammad Saleem", "Information Technology")

# Display information using the display method
teacher1.display()
```

9. Abstract Classes and Methods
Assignment:
Use the abc module to create an abstract class Shape with an abstract method area(). Inherit a class Rectangle that implements area().**bold text**

```python
from abc import ABC, abstractmethod  # Importing ABC and abstractmethod

# Abstract class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass  # Abstract method, doesn't implement anything here

# Derived class Rectangle that implements area()
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height  # Implementing the area method

# Create an object of Rectangle
rectangle = Rectangle(5, 10)

# Call the area method
print("Area of rectangle:", rectangle.area())

```

14. Aggregation
Assignment:
Create a class Department and a class Employee. Use aggregation by having a Department object store a reference to an Employee object that exists independently of **it**

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def get_details(self):
        return f"Employee Name: {self.name}"

class Department:
    def __init__(self, dept_name, employee):
        self.dept_name = dept_name
        self.employee = employee  # Aggregation: Storing reference to existing Employee object

    def show_department_info(self):
        return f"Department: {self.dept_name}, {self.employee.get_details()}"

# Create an Employee object (exists independently)
emp = Employee("Muhammad Saleem")

# Create a Department object and pass the Employee object
dept = Department("HR", emp)

# Show information
print(dept.show_department_info())
```

15. Method Resolution Order (MRO) and Diamond Inheritance
Assignment:
Create four classes:

A with a method show(),

B and C that inherit from A and override show(),

D that inherits from both B and C.

Create an object of D and call show() to observe MRO.**bold text**

```python
class A:
    def show(self):
        print("Show from class A")

class B(A):
    def show(self):
        print("Show from class B")

class C(A):
    def show(self):
        print("Show from class C")

class D(B, C):  # Diamond inheritance
    pass

# Create an object of D and call show
d = D()
d.show()

# Print MRO
print("MRO:", [cls.__name__ for cls in D.__mro__])
```

16. Function Decorators
Assignment:
Write a decorator function log_function_call that prints "Function is being called" before a function executes. Apply it to a function say_hello().

**bold text**

```python
# Decorator function
def log_function_call(func):
    def wrapper():
        print("Function is being called")
        return func()
    return wrapper

# Function with decorator applied
@log_function_call
def say_hello():
    print("Hello!")

# Call the function
say_hello()
```

17. Class Decorators
Assignment:
Create a class decorator add_greeting that modifies a class to add a greet() method returning "Hello from Decorator!". Apply it to a class Person.**bold text**

```python
# Class decorator
def add_greeting(cls):
    def greet(self):
        return "Hello from Decorator!"

    cls.greet = greet  # Dynamically add greet() method to the class
    return cls

# Apply the decorator to Person class
@add_greeting
class Person:
    def __init__(self, name):
        self.name = name

# Create an object of Person and call greet()
p = Person("Alice")
print(p.greet())
```

18. Property Decorators: @property, @setter, and @deleter
Assignment:
Create a class Product with a private attribute _price. Use @property to get the price, @price.setter to update it, and @price.deleter to delete it.**bold text**

```python
class Product:
    def __init__(self, price):
        self._price = price  # Private attribute

    # Getter method using @property
    @property
    def price(self):
        print("Getting the price...")
        return self._price

    # Setter method using @price.setter
    @price.setter
    def price(self, value):
        if value < 0:
            print("Price cannot be negative!")
        else:
            print("Setting the price...")
            self._price = value

    # Deleter method using @price.deleter
    @price.deleter
    def price(self):
        print("Deleting the price...")
        del self._price

# Example usage
p = Product(100)

# Getting the price
print(p.price)  # Calls the @property getter

# Setting a new price
p.price = 150  # Calls the @price.setter

# Trying to set a negative price (won't work)
p.price = -50  # Prints the error message

# Deleting the price
del p.price  # Calls the @price.deleter

# Trying to get the price after deletion (will raise an error)
# print(p.price)  # Uncommenting this will raise an AttributeError because _price was deleted
```

19. callable() and __call__()
Assignment:
Create a class Multiplier with an __init__() to set a factor. Define a __call__() method that multiplies an input by the factor. Test it with callable() and by calling the object like a function.**bold text**

```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return value * self.factor

# Create an object with a multiplication factor of 3
triple = Multiplier(3)

# Check if the object is callable
print(callable(triple))  # Output: True

# Use the object like a function
print(triple(10))  # Output: 30
print(triple(7))   # Output: 21
```

20. Creating a Custom Exception
Assignment:
Create a custom exception InvalidAgeError. Write a function check_age(age) that raises this exception if age < 18. Handle it with try...except.**bold text**

```python
# Custom Exception
class InvalidAgeError(Exception):
    pass  # You can add a custom message if you want

# Function that checks age
def check_age(age):
    if age < 18:
        raise InvalidAgeError("Age must be at least 18.")
    else:
        print("Access granted.")

# Handling with try...except
try:
    age = int(input("Enter your age: "))
    check_age(age)
except InvalidAgeError as e:
    print("InvalidAgeError caught:", e)
except ValueError:
    print("Please enter a valid number.")
```

21. Make a Custom Class Iterable
Assignment:
Create a class Countdown that takes a start number. Implement __iter__() and __next__() to make the object iterable in a for-loop, counting down to 0.**bold text**

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self  # Returns the iterator object itself

    def __next__(self):
        if self.current < 0:
            raise StopIteration  # Stop when below 0
        else:
            number = self.current
            self.current -= 1
            return number

# Example usage
for num in Countdown(5):
    print(num)
```
